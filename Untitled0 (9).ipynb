```python
import os, gc, glob, random, time, numpy as np, torch, torch.nn as nn, torch.nn.functional as F
from torchvision import models
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt

IMG_SIZE, BATCH, EPOCHS, LR, SEED = 96, 128, 30, 2e-3, 42   # 96px keeps Colab RAM safe
MODEL_FN = models.resnet18   # random weights (not pre-trained)

random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

```python
import tensorflow as tf
tf.config.set_visible_devices([], 'GPU')   # TF only decodes images; GPU stays free for PyTorch

feature_spec = {
    'image': tf.io.FixedLenFeature([], tf.string),
    'super_class_id': tf.io.FixedLenFeature([], tf.int64),
    'class_id': tf.io.FixedLenFeature([], tf.int64),
}
def parse(rec):
    ex = tf.io.parse_single_example(rec, feature_spec)
    img = tf.io.decode_image(ex['image'], channels=3, expand_animations=False)
    img = tf.image.resize(img, (IMG_SIZE, IMG_SIZE), antialias=True)
    img = tf.cast(tf.round(tf.clip_by_value(img, 0, 255)), tf.uint8)
    return img, ex['super_class_id'], ex['class_id']

def load(pattern):
    files = sorted(glob.glob(pattern, recursive=True))
    print(len(files), "files for", pattern)
    ds = (tf.data.TFRecordDataset(files, num_parallel_reads=tf.data.AUTOTUNE)
          .map(parse, num_parallel_calls=tf.data.AUTOTUNE).batch(1024).prefetch(2))
    X, Y, G = [], [], []
    for x, y, g in ds:
        X.append(x.numpy()); Y.append(y.numpy()); G.append(g.numpy())
    X = np.concatenate(X); gc.collect()
    return X, np.concatenate(Y), np.concatenate(G)

t0 = time.time()
X_trval, y_trval, g_trval = load(path + '/**/*-train.tfrecord*')
X_test,  y_test,  _       = load(path + '/**/*-test.tfrecord*')
print("Loaded in", round(time.time() - t0), "s |", X_trval.shape, X_test.shape)

uniq = np.unique(y_trval)
y_trval = np.searchsorted(uniq, y_trval); y_test = np.searchsorted(uniq, y_test)
SOP_NAMES = ['bicycle','cabinet','chair','coffee_maker','fan','kettle','lamp','mug','sofa','stapler','table','toaster']
classes = SOP_NAMES if len(uniq) == 12 else [str(u) for u in uniq]
print("Super-classes:", len(classes), classes)
```

```python
gss = GroupShuffleSplit(n_splits=1, test_size=0.10, random_state=SEED)
tr_idx, va_idx = next(gss.split(X_trval, y_trval, groups=g_trval))   # no product appears in both

def to_tensor(X): return torch.from_numpy(X).permute(0, 3, 1, 2).contiguous()   # uint8, NCHW
Xtr, ytr = to_tensor(X_trval[tr_idx]).to(device), torch.from_numpy(y_trval[tr_idx]).to(device)
Xva, yva = to_tensor(X_trval[va_idx]).to(device), torch.from_numpy(y_trval[va_idx]).to(device)
Xte, yte = to_tensor(X_test), torch.from_numpy(y_test)          # test stays on CPU
del X_trval, X_test
gc.collect()
print("train", len(ytr), "| val", len(yva), "| test", len(yte))

mean = torch.tensor([0.485, 0.456, 0.406], device=device).view(1, 3, 1, 1)
std  = torch.tensor([0.229, 0.224, 0.225], device=device).view(1, 3, 1, 1)
def normalize(xb): return (xb.float() / 255 - mean) / std

def augment(xb):   # fast GPU augmentation: rotate, zoom, shift, flip, brightness, contrast
    x = xb.float() / 255; B = x.size(0)
    ang  = (torch.rand(B, device=device) - 0.5) * 2 * (15 * np.pi / 180)
    sc   = 1 / (0.8 + 0.4 * torch.rand(B, device=device))
    tx   = (torch.rand(B, device=device) - 0.5) * 0.2
    ty   = (torch.rand(B, device=device) - 0.5) * 0.2
    flip = (torch.rand(B, device=device) < 0.5).float() * 2 - 1
    cos, sin = torch.cos(ang) * sc, torch.sin(ang) * sc
    theta = torch.stack([torch.stack([cos * flip, -sin, tx], 1),
                         torch.stack([sin * flip,  cos, ty], 1)], 1)
    grid = F.affine_grid(theta, x.shape, align_corners=False)
    x = F.grid_sample(x, grid, padding_mode="border", align_corners=False)
    b = 1 + (torch.rand(B, 1, 1, 1, device=device) - 0.5) * 0.4
    c = 1 + (torch.rand(B, 1, 1, 1, device=device) - 0.5) * 0.4
    m = x.mean((1, 2, 3), keepdim=True)
    x = (((x - m) * c + m) * b).clamp(0, 1)
    return (x - mean) / std

def get_batches(X, y, bs, shuffle=False, aug=False, drop_last=False):
    n = len(y); order = torch.randperm(n) if shuffle else torch.arange(n)
    for i in range(0, n, bs):
        ids = order[i:i + bs]
        if drop_last and len(ids) < bs: break
        xb = X[ids.to(X.device)].to(device, non_blocking=True)
        yb = y[ids.to(y.device)].to(device, non_blocking=True)
        yield (augment(xb) if aug else normalize(xb)), yb
```

```python
import copy
ARCHS = {                                   # all trained from random initialization
    "resnet18":        models.resnet18,
    "resnet34":        models.resnet34,
    "regnet_y_800mf":  models.regnet_y_800mf,
    "efficientnet_b0": models.efficientnet_b0,
}
EPOCHS = 30
criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

def run_epoch(model, X, y, train, opt=None, sched=None, scaler=None):
    model.train(train)
    loss_sum, correct, n = 0.0, 0, 0
    with torch.set_grad_enabled(train):
        for xb, yb in get_batches(X, y, BATCH, shuffle=train, aug=train, drop_last=train):
            with torch.autocast(device_type="cuda", enabled=device.type == "cuda"):
                out = model(xb); loss = criterion(out, yb)
            if train:
                opt.zero_grad(set_to_none=True)
                scaler.scale(loss).backward()
                scaler.step(opt); scaler.update(); sched.step()
            loss_sum += loss.item() * len(yb)
            correct += (out.argmax(1) == yb).sum().item(); n += len(yb)
    return loss_sum / n, correct / n

def train_one(name, fn, epochs=EPOCHS):
    torch.manual_seed(SEED)
    model = fn(weights=None, num_classes=len(classes)).to(device)
    opt = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-2)
    sched = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=LR, epochs=epochs, steps_per_epoch=len(ytr) // BATCH)
    scaler = torch.amp.GradScaler("cuda", enabled=device.type == "cuda")
    hist = {"tr_acc": [], "va_acc": [], "tr_loss": [], "va_loss": []}
    best, best_state = 0, None
    for ep in range(epochs):
        t0 = time.time()
        trl, tra = run_epoch(model, Xtr, ytr, True, opt, sched, scaler)
        val, vaa = run_epoch(model, Xva, yva, False)
        for k, v in zip(hist, (tra, vaa, trl, val)): hist[k].append(v)
        if vaa > best:
            best = vaa; best_state = copy.deepcopy(model.state_dict()); torch.save(best_state, f"best_{name}.pth")
        print(f"[{name}] epoch {ep+1:02d}/{epochs} | train {tra:.4f} | val {vaa:.4f} | best {best:.4f} | {time.time()-t0:.0f}s")
    model.load_state_dict(best_state)
    return model, hist, best
```

```python
import os, gc, glob, random, time, copy, numpy as np, torch, torch.nn as nn, torch.nn.functional as F
from torchvision import models
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt

IMG_SIZE, BATCH, EPOCHS, LR, SEED = 112, 128, 60, 2e-3, 42   # use 96 if Colab runs out of RAM

random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

```python
ARCHS = {"resnet18": models.resnet18, "resnet34": models.resnet34, "regnet_y_800mf": models.regnet_y_800mf}
criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
CUTMIX_P, EMA_DECAY = 0.6, 0.998

def cutmix(xb, yb):
    lam = np.random.beta(1.0, 1.0); perm = torch.randperm(xb.size(0), device=xb.device)
    H, W = xb.shape[2:]; r = np.sqrt(1 - lam)
    ch, cw = int(H * r), int(W * r); cy, cx = np.random.randint(H), np.random.randint(W)
    y1, y2 = max(cy - ch // 2, 0), min(cy + ch // 2, H); x1, x2 = max(cx - cw // 2, 0), min(cx + cw // 2, W)
    xb = xb.clone(); xb[:, :, y1:y2, x1:x2] = xb[perm][:, :, y1:y2, x1:x2]
    lam = 1 - ((y2 - y1) * (x2 - x1) / (H * W))
    return xb, yb, yb[perm], lam

@torch.no_grad()
def ema_update(ema, model, d=EMA_DECAY):
    for pe, p in zip(ema.parameters(), model.parameters()): pe.mul_(d).add_(p.detach(), alpha=1 - d)
    for be, b in zip(ema.buffers(), model.buffers()): be.copy_(b)

def run_epoch(model, X, y, train, opt=None, sched=None, scaler=None, ema=None):
    model.train(train)
    loss_sum, correct, n = 0.0, 0, 0
    with torch.set_grad_enabled(train):
        for xb, yb in get_batches(X, y, BATCH, shuffle=train, aug=train, drop_last=train):
            xb = xb.contiguous(memory_format=torch.channels_last)
            with torch.autocast(device_type="cuda", enabled=device.type == "cuda"):
                if train and np.random.rand() < CUTMIX_P:
                    xb, ya, yb2, lam = cutmix(xb, yb)
                    out = model(xb); loss = lam * criterion(out, ya) + (1 - lam) * criterion(out, yb2)
                else:
                    out = model(xb); loss = criterion(out, yb)
            if train:
                opt.zero_grad(set_to_none=True)
                scaler.scale(loss).backward(); scaler.step(opt); scaler.update(); sched.step()
                ema_update(ema, model)
            loss_sum += loss.item() * len(yb)
            correct += (out.argmax(1) == yb).sum().item(); n += len(yb)
    return loss_sum / n, correct / n

def train_one(name, fn, epochs=EPOCHS):
    torch.manual_seed(SEED)
    model = fn(weights=None, num_classes=len(classes)).to(device).to(memory_format=torch.channels_last)
    ema = copy.deepcopy(model)
    opt = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=5e-2)
    sched = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=LR, epochs=epochs,
                                                steps_per_epoch=len(ytr) // BATCH, pct_start=0.2)
    scaler = torch.amp.GradScaler("cuda", enabled=device.type == "cuda")
    hist = {"va_acc": [], "va_loss": []}; best = 0
    for ep in range(epochs):
        t0 = time.time()
        trl, _ = run_epoch(model, Xtr, ytr, True, opt, sched, scaler, ema)
        val, vaa = run_epoch(ema, Xva, yva, False)
        hist["va_acc"].append(vaa); hist["va_loss"].append(val)
        if vaa > best: best = vaa; torch.save(ema.state_dict(), f"best_{name}.pth")
        print(f"[{name}] ep {ep+1:02d}/{epochs} | loss {trl:.3f} | val {vaa:.4f} | best {best:.4f} | {time.time()-t0:.0f}s")
    ema.load_state_dict(torch.load(f"best_{name}.pth"))
    return ema, hist, best
```

```python
trained, histories, val_scores = {}, {}, {}
for name, fn in ARCHS.items():
    print(f"\n===== Training {name} from scratch =====")
    trained[name], histories[name], val_scores[name] = train_one(name, fn)
    gc.collect(); torch.cuda.empty_cache()
print("Best val acc:", {k: round(v, 4) for k, v in val_scores.items()})
```

```python
# ============================================================
# LOST & FOUND + MULTIMODAL PRODUCT CLASSIFIER
# Google Colab | One app cell after your model setup/training
# ============================================================

!pip -q install gradio

import os
import csv
import uuid
import datetime
import numpy as np
import gradio as gr
import torch
import torch.nn.functional as F

from PIL import Image
from google.colab import drive

# ---------------------- GOOGLE DRIVE ------------------------

drive.mount("/content/drive")

DRIVE_ROOT = "/content/drive/MyDrive/LostAndFoundUploads"
LOST_DIR = os.path.join(DRIVE_ROOT, "Lost")
FOUND_DIR = os.path.join(DRIVE_ROOT, "Found")

os.makedirs(LOST_DIR, exist_ok=True)
os.makedirs(FOUND_DIR, exist_ok=True)

CSV_PATH = os.path.join(DRIVE_ROOT, "reports.csv")
CSV_COLS = [
    "id", "type", "time", "name", "contact",
    "location", "description", "category", "image"
]

if not os.path.exists(CSV_PATH):
    with open(CSV_PATH, "w", newline="", encoding="utf-8") as f:
        csv.writer(f).writerow(CSV_COLS)

# ---------------------- MODEL SETUP -------------------------

# Uses your existing trained checkpoints and ARCHS dictionary.
demo_models = {}

for n, fn in ARCHS.items():
    model = fn(weights=None, num_classes=len(classes)).to(device)
    model = model.to(memory_format=torch.channels_last)
    model.load_state_dict(
        torch.load(f"best_{n}.pth", map_location=device)
    )
    model.eval()
    demo_models[n] = model

def cl(t):
    return t.contiguous(memory_format=torch.channels_last)

# ---------------------- APPEARANCE FEATURES -----------------

def _backbone_feats(x):
    model = demo_models["resnet34"]

    x = model.maxpool(
        model.relu(model.bn1(model.conv1(x)))
    )

    x = model.layer2(model.layer1(x))
    l3 = model.layer3(x)
    l4 = model.layer4(l3)

    return torch.cat([
        F.normalize(l3.float().mean((2, 3)), dim=1),
        F.normalize(l4.float().mean((2, 3)), dim=1)
    ], dim=1)


@torch.no_grad()
def appearance(x8):
    # x8: uint8 tensor, shape (B, 3, H, W)

    with torch.autocast(
        device_type="cuda",
        enabled=device.type == "cuda"
    ):
        deep = F.normalize(
            _backbone_feats(cl(normalize(x8))).float(),
            dim=1
        )

    # Colour histogram
    q = x8.long() // 64
    idx = (
        q[:, 0] * 16 + q[:, 1] * 4 + q[:, 2]
    ).flatten(1)

    hist = torch.zeros(
        len(x8), 64, device=x8.device
    ).scatter_add_(
        1,
        idx,
        torch.ones(idx.shape, device=x8.device)
    )

    colour = F.normalize(hist.sqrt(), dim=1)

    # Spatial colour/layout representation
    layout = F.adaptive_avg_pool2d(
        x8.float() / 255, 12
    ).flatten(1)

    layout = F.normalize(
        layout - layout.mean(1, keepdim=True),
        dim=1
    )

    return deep, colour, layout


APP_W = (0.5, 0.3, 0.2)


def app_sim(q, gallery_features):
    return sum(
        weight * (gallery_part @ query_part)
        for weight, query_part, gallery_part
        in zip(APP_W, q, gallery_features)
    )


# Precompute appearance features for training images
with torch.no_grad():
    parts = [
        appearance(Xtr[i:i + 512])
        for i in range(0, len(Xtr), 512)
    ]

    gallery = tuple(
        torch.cat([part[k] for part in parts])
        for k in range(3)
    )

print("Training-image gallery ready:", [
    tuple(g.shape) for g in gallery
])


# ---------------------- IMAGE PREPROCESSING -----------------

def prep(img):
    img = img.convert("RGB").resize(
        (IMG_SIZE, IMG_SIZE), Image.BILINEAR
    )

    return torch.from_numpy(
        np.array(img)
    ).permute(2, 0, 1).unsqueeze(0).to(device)


# ---------------------- SHARED CLASSIFIER -------------------

@torch.no_grad()
def classify(img, description="", img_weight=0.6):
    x8 = prep(img)
    x = cl(normalize(x8))

    probabilities = []

    for model in demo_models.values():
        probabilities.append(model(x).softmax(1))
        probabilities.append(model(x.flip(3)).softmax(1))

    p_img = (
        sum(probabilities) / len(probabilities)
    )[0].float().cpu().numpy()

    q = tuple(t[0] for t in appearance(x8))
    qf = tuple(t[0] for t in appearance(x8.flip(3)))
    emb = (q, qf)

    if description and description.strip():
        p_txt, cos = text_scores(description)
        fused = (
            img_weight * p_img
            + (1 - img_weight) * p_txt
        )
    else:
        p_txt, cos = None, None
        fused = p_img.copy()

    fused = fused / fused.sum()

    return p_img, p_txt, cos, fused, emb


# ---------------------- CLASSIFIER TAB ----------------------

@torch.no_grad()
def predict(img, description, img_weight):
    if img is None:
        return None, None, None, []

    try:
        p_img, p_txt, cos, fused, emb = classify(
            img, description, img_weight
        )

        out_img = {
            classes[i]: float(p_img[i])
            for i in range(len(classes))
        }

        q, qf = emb
        similarities = torch.maximum(
            app_sim(q, gallery),
            app_sim(qf, gallery)
        )

        top = similarities.topk(min(6, len(similarities)))
        similar_images = []

        for score, idx in zip(
            top.values.tolist(),
            top.indices.tolist()
        ):
            im = Image.fromarray(
                Xtr[idx].permute(1, 2, 0).cpu().numpy()
            ).resize((224, 224), Image.BICUBIC)

            similar_images.append((
                im,
                f"{classes[ytr[idx].item()]} | similarity {score:.2f}"
            ))

        out_txt = (
            {
                classes[i]: float(cos[i])
                for i in range(len(classes))
            }
            if cos is not None
            else {"No description given": 1.0}
        )

        out_fused = {
            classes[i]: float(fused[i])
            for i in range(len(classes))
        }

        return out_img, out_txt, out_fused, similar_images

    except Exception as e:
        import traceback
        traceback.print_exc()
        raise gr.Error(f"{type(e).__name__}: {e}")


# ---------------------- REPORT STORAGE ----------------------

REPORTS = []

def text_sim(a, b):
    if a is None or b is None:
        return None

    return float(
        sk_norm(a).multiply(sk_norm(b)).sum()
    )


def make_embedding(img, description):
    # Recreate the visual embedding from the saved image.
    _, _, _, _, (emb, _) = classify(img, description)
    return emb


def restore_reports():
    """Reload saved reports and rebuild embeddings after restart."""
    REPORTS.clear()

    if not os.path.exists(CSV_PATH):
        return

    with open(CSV_PATH, "r", newline="", encoding="utf-8") as f:
        rows = list(csv.DictReader(f))

    for item in rows:
        image_path = item.get("image", "")
        if not image_path or not os.path.exists(image_path):
            continue

        try:
            desc = item.get("description", "")
            img = Image.open(image_path).convert("RGB")
            emb = make_embedding(img, desc)

            REPORTS.append({
                "id": item.get("id", ""),
                "type": item.get("type", ""),
                "time": item.get("time", ""),
                "name": item.get("name", ""),
                "contact": item.get("contact", ""),
                "location": item.get("location", ""),
                "description": desc,
                "category": item.get("category", ""),
                "image": image_path,
                "emb": emb,
                "vec": tfidf.transform([desc]) if desc.strip() else None
            })

        except Exception as e:
            print("Skipped report:", item.get("id"), e)

    print(f"Restored {len(REPORTS)} saved report(s).")


def find_matches(rec, k=5):
    scored = []

    for r in REPORTS:
        # Match lost reports only with found reports and vice versa.
        if r["type"] == rec["type"]:
            continue

        img_sim = max(
            0.0,
            min(
                1.0,
                sum(
                    weight * float(a @ b)
                    for weight, a, b
                    in zip(APP_W, rec["emb"], r["emb"])
                )
            )
        )

        same_cat = float(
            rec["category"] == r["category"]
        )

        t = text_sim(rec["vec"], r["vec"])

        if t is not None:
            score = (
                0.5 * img_sim
                + 0.3 * t
                + 0.2 * same_cat
            )
        else:
            score = (
                0.7 * img_sim
                + 0.3 * same_cat
            )

        scored.append((score, r))

    scored.sort(key=lambda x: x[0], reverse=True)
    return scored[:k]


HEADERS = [
    "Match %", "Type", "Category", "Location",
    "Contact", "Name", "Description", "Reported at"
]


def row(r, score=None):
    return [
        f"{score * 100:.0f}%" if score is not None else "",
        r["type"].upper(),
        r["category"],
        r["location"],
        r["contact"],
        r["name"],
        r["description"],
        r["time"]
    ]


# ---------------------- SUBMIT LOST/FOUND REPORT ------------

def submit_report(
    kind, img, name, location, contact, desc, manual_category
):
    if img is None:
        raise gr.Error("Please upload a photo of the item.")

    if not (location or "").strip():
        raise gr.Error("Please enter the location.")

    if not (contact or "").strip():
        raise gr.Error("Please enter your contact details.")

    desc = (desc or "").strip()
    manual_category = manual_category or ""

    try:
        _, _, _, fused, (emb, _) = classify(img, desc)

        # Blank category means use the AI's prediction.
        if manual_category:
            cat = manual_category
        else:
            cat = classes[int(fused.argmax())]

        rid = uuid.uuid4().hex[:10]
        folder = LOST_DIR if kind == "lost" else FOUND_DIR

        path = os.path.join(folder, f"{kind}_{rid}.png")
        img.convert("RGB").save(path)

        rec = {
            "id": rid,
            "type": kind,
            "time": datetime.datetime.now().strftime(
                "%Y-%m-%d %H:%M"
            ),
            "name": (name or "").strip() or "(not given)",
            "contact": contact.strip(),
            "location": location.strip(),
            "description": desc,
            "category": cat,
            "image": path,
            "emb": emb,
            "vec": tfidf.transform([desc]) if desc else None
        }

        # Find matches against previously submitted opposite-type reports.
        matches = find_matches(rec)

        # Store the report for future matching.
        REPORTS.append(rec)

        with open(
            CSV_PATH, "a", newline="", encoding="utf-8"
        ) as f:
            writer = csv.writer(f)
            writer.writerow([rec[c] for c in CSV_COLS])

    except gr.Error:
        raise
    except Exception as e:
        import traceback
        traceback.print_exc()
        raise gr.Error(f"{type(e).__name__}: {e}")

    status = (
        f"**Report saved! ID: {rid}**\n\n"
        f"Category: **{cat}**\n\n"
        f"Photo saved to Google Drive: `{folder}`\n\n"
    )

    if matches:
        status += f"Found {len(matches)} possible match(es) below."
    else:
        opposite = "found" if kind == "lost" else "lost"
        status += (
            f"No matching {opposite} reports found yet. "
            "The report is saved and can be matched against future reports."
        )

    # Show the uploaded item image in the report results.
    report_image = img.convert("RGB")

    # Keep category probabilities visible, but highlight the selected category.
    cat_output = {
        classes[i]: float(fused[i])
        for i in range(len(classes))
    }

    if matches:
        match_gallery = []

        for score, match in matches:
            try:
                match_img = Image.open(
                    match["image"]
                ).convert("RGB").resize((224, 224))

                caption = (
                    f"{score * 100:.0f}% | {match['category']} | "
                    f"{match['location']} | {match['contact']}"
                )

                match_gallery.append((match_img, caption))

            except Exception:
                continue

        match_table = [row(r, score) for score, r in matches]
    else:
        match_gallery = []
        match_table = []

    return status, report_image, cat_output, match_gallery, match_table


# ---------------------- ALL REPORTS -------------------------

def list_reports(which):
    selected = [
        r for r in REPORTS
        if which == "All" or r["type"] == which.lower()
    ]

    return [row(r) for r in reversed(selected)]


def report_gallery(which):
    selected = [
        r for r in REPORTS
        if which == "All" or r["type"] == which.lower()
    ]

    images = []

    for r in reversed(selected):
        if os.path.exists(r["image"]):
            images.append((
                Image.open(r["image"]).convert("RGB"),
                f"{r['type'].upper()} | {r['category']} | {r['location']}"
            ))

    return images


# ---------------------- EXAMPLE PRODUCTS --------------------

os.makedirs("examples", exist_ok=True)

DESC = {
    "bicycle": "black mountain bike with gears, pedals and handlebars",
    "cabinet": "wooden storage cabinet with doors and shelves",
    "chair": "modern office chair with backrest and wheels",
    "coffee_maker": "drip coffee maker with water tank and carafe",
    "fan": "white pedestal fan with blades and speed settings",
    "kettle": "electric kettle with spout and auto shut off",
    "lamp": "silver desk lamp with adjustable arm and LED light",
    "mug": "ceramic coffee mug with handle",
    "sofa": "grey fabric sofa with cushions and armrests",
    "stapler": "office stapler with metal spring and staples",
    "table": "wooden dining table with four legs",
    "toaster": "two slice toaster with browning dial and lever"
}

examples, labels = [], []
yte_np = yte.cpu().numpy() if torch.is_tensor(yte) else np.asarray(yte)

for ci, cname in enumerate(classes):
    idx = np.where(yte_np == ci)[0]

    if len(idx) == 0:
        continue

    fp = f"examples/{cname}.png"

    Image.fromarray(
        Xte[idx[0]].permute(1, 2, 0).cpu().numpy()
    ).resize((224, 224), Image.BICUBIC).save(fp)

    examples.append([fp, DESC.get(cname, ""), 0.6])
    labels.append(cname)


# Demo reports use fake contact details.
lost_ex = []
found_ex = []

if len(examples) >= 8:
    lost_ex = [
        [
            examples[5][0], "Aarav", "Library, Block B",
            "aarav@example.com",
            "black electric kettle with spout", None
        ],
        [
            examples[7][0], "Meera", "Canteen",
            "98XXXXXX01",
            "ceramic coffee mug with handle", None
        ]
    ]

    found_ex = [
        [
            examples[5][0], "Rohan", "Hostel gate",
            "98XXXXXX02",
            "electric kettle with spout, left near the gate", None
        ],
        [
            examples[7][0], "Sana", "Canteen counter",
            "sana@example.com",
            "white ceramic mug", None
        ]
    ]


# ---------------------- RESTORE SAVED REPORTS ----------------

restore_reports()


# ---------------------- GRADIO APP ----------------------------

with gr.Blocks(
    title="Lost & Found Product Classifier"
) as demo:

    gr.Markdown(
        "# Lost & Found + Multimodal Product Classifier\n"
        "CNN ensemble + image appearance similarity + TF-IDF text matching.\n\n"
        "Reports and images are saved in Google Drive."
    )

    with gr.Tabs():

        # ---------------- CLASSIFIER TAB ---------------------
        with gr.Tab("Classifier"):

            with gr.Row():
                with gr.Column():
                    c_img = gr.Image(
                        type="pil",
                        label="Product photo",
                        height=300
                    )

                    c_desc = gr.Textbox(
                        label="Product description (optional)",
                        placeholder="e.g. black electric kettle with spout"
                    )

                    c_w = gr.Slider(
                        0, 1, value=0.6, step=0.05,
                        label="Weight on image vs description"
                    )

                    c_btn = gr.Button(
                        "Classify item",
                        variant="primary"
                    )

                with gr.Column():
                    c_o1 = gr.Label(
                        num_top_classes=3,
                        label="Image-based prediction"
                    )

                    c_o2 = gr.Label(
                        num_top_classes=3,
                        label="Description similarity"
                    )

                    c_o3 = gr.Label(
                        num_top_classes=3,
                        label="Combined prediction"
                    )

            c_gal = gr.Gallery(
                label="Most similar training images",
                columns=6,
                rows=1,
                height=260,
                object_fit="contain"
            )

            c_btn.click(
                predict,
                [c_img, c_desc, c_w],
                [c_o1, c_o2, c_o3, c_gal]
            )

            if examples:
                gr.Examples(
                    examples=examples,
                    inputs=[c_img, c_desc, c_w],
                    outputs=[c_o1, c_o2, c_o3, c_gal],
                    fn=predict,
                    cache_examples=False,
                    run_on_click=True
                )

        # ---------------- LOST / FOUND TABS ------------------
        for kind, title, ex in [
            ("lost", "I lost an item", lost_ex),
            ("found", "I found an item", found_ex)
        ]:

            with gr.Tab(title):

                opposite = "found" if kind == "lost" else "lost"

                gr.Markdown(
                    f"### Report a {kind} item\n"
                    f"Your report will be matched against {opposite} reports."
                )

                with gr.Row():

                    with gr.Column():

                        r_img = gr.Image(
                            type="pil",
                            label=f"Photo of the {kind} item",
                            height=280
                        )

                        r_name = gr.Textbox(
                            label="Your name"
                        )

                        r_loc = gr.Textbox(
                            label=f"Location where it was {kind}",
                            placeholder="e.g. Library, Block B"
                        )

                        r_con = gr.Textbox(
                            label="Your contact (phone / email)"
                        )

                        r_desc = gr.Textbox(
                            label="Description (colour, brand, marks...)",
                            lines=2
                        )

                        # No default category selected.
                        # Leave blank to use the AI prediction.
                        r_manual_cat = gr.Dropdown(
                            choices=list(classes),
                            value=None,
                            label="Category (optional — blank uses AI prediction)",
                            allow_custom_value=False
                        )

                        r_btn = gr.Button(
                            f"Submit {kind} report",
                            variant="primary"
                        )

                    with gr.Column():
                        r_stat = gr.Markdown()

                        r_uploaded_img = gr.Image(
                            type="pil",
                            label="Your submitted item photo",
                            height=220
                        )

                        r_cat = gr.Label(
                            num_top_classes=3,
                            label="Category prediction"
                        )

                r_gal = gr.Gallery(
                    label="Possible matches",
                    columns=5,
                    rows=1,
                    height=260,
                    object_fit="contain"
                )

                r_tab = gr.Dataframe(
                    headers=HEADERS,
                    label="Match details",
                    wrap=True
                )

                if ex:
                    gr.Examples(
                        examples=ex,
                        inputs=[
                            r_img, r_name, r_loc, r_con,
                            r_desc, r_manual_cat
                        ],
                        label="Fill with an example"
                    )

                r_btn.click(
                    fn=lambda img, name, loc, con, desc, cat, k=kind:
                        submit_report(k, img, name, loc, con, desc, cat),
                    inputs=[
                        r_img, r_name, r_loc, r_con,
                        r_desc, r_manual_cat
                    ],
                    outputs=[
                        r_stat, r_uploaded_img, r_cat,
                        r_gal, r_tab
                    ]
                )

        # ---------------- ALL REPORTS TAB --------------------
        with gr.Tab("All Reports"):

            b_f = gr.Radio(
                ["All", "Lost", "Found"],
                value="All",
                label="Show reports"
            )

            b_t = gr.Dataframe(
                headers=HEADERS,
                label="Report records",
                wrap=True
            )

            b_gal = gr.Gallery(
                label="Reported item photos",
                columns=5,
                height=320,
                object_fit="contain"
            )

            refresh_btn = gr.Button("Refresh reports")

            refresh_btn.click(
                fn=lambda selected: (
                    list_reports(selected),
                    report_gallery(selected)
                ),
                inputs=b_f,
                outputs=[b_t, b_gal]
            )

            b_f.change(
                fn=lambda selected: (
                    list_reports(selected),
                    report_gallery(selected)
                ),
                inputs=b_f,
                outputs=[b_t, b_gal]
            )


# ---------------------- LAUNCH APP ---------------------------

demo.launch(share=True, debug=True)
```
